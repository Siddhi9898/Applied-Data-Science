```python
######## Object Oriented programming, part II, create your own objects ###############

# libraries 
import pickle

# Remember the Mailcat example from the Dictionary video lecture?, Let's re-use the Mailcat.
class Mailcat_object():
    def __init__(self, id_name, mail_adress, list_items=[]):  # Creates an instance of this class
        self.idnum: int  # idnumber 
        self.id_name = id_name  # name 
        self.mail_adress = mail_adress # mail data 
        self.list_items = list_items # arbitrary data items in list

    def get_id_name(self):  # These methods only shows how to create methods.
        return self.id_name  # Returns string
    def get_mail_adress(self): 
        return self.mail_adress  # Returns string
    def get_list_items(self):
        return self.list_items # Returns a list of arbitrary objects

# Contains a class "Mailcat_collection" with a dictionary, using idnum as key.
class Mailcat_collection():
    def __init__(self):
        self.Mailcat = {} # Dictionary with idnum as key and Mailcat_objects as values
        self.index_counter = 0 # The first key number

    def addMailcat_object(self, Mailcat_object):
        self.index_counter += 1 # create the next key-number to be used in the catalogue
        Mailcat_collection.idnum = self.index_counter # key-number
        self.Mailcat.update({Mailcat_collection.idnum: Mailcat_object}) # This line adds the Mailcat_object to the Mailcat_collection

# Create a class which *encapsulates* our Mailcatalogue and adds four start entries into the Mailcat_collection object named "data"
# The class has a function for printing the contents of the Mailcatalogue.
class create_catalog():
    def __init__(self):
        self.data = Mailcat_collection()
        self.data.addMailcat_object(Mailcat_object("Joe", "joe.joe@gmail.com", [1200, 99.3, 0, 0.0, "value", "compl%"]))
        self.data.addMailcat_object(Mailcat_object('Anand', "anand.anand@aol.com", [1300, 75.3, 4, 33.3, "value", "compl%", "Time_c", "answ%"]))
        self.data.addMailcat_object(Mailcat_object('Lisa', "lisa@gmx.com", [2400, 45.4, 0, 0.0, "value", "compl%"]))
        self.data.addMailcat_object(Mailcat_object('Mohammed', "mohammed.gmail.com", [1700, 82.3, 0, 0.0, "value", "compl%"]))

    def list_Mailcat_objects(self):
        dict = self.data.Mailcat
        for key in dict:
            mail= dict[key] # get the value (object instance) and print the relevant data values and objects
            print (mail.id_name, mail.mail_adress, mail.list_items)

    def serialize_Mailcat_objects(self):
        dict = self.data.Mailcat
        Ret_list = []
        Header = ["Name", "mail_adress", "integer1", "float1", "integer2", "float2", "string1", "string2", "string3", "string4"]
        Ret_list.append(Header)
        for key in dict:
            mail= dict[key] # get the value (object instance) and print the relevant data values and objects
            list_string = ' '.join([str(item) for item in mail.list_items])
            mail_string = str(mail.id_name) + " " + str(mail.mail_adress) + " " + list_string
            Ret_list.append(mail_string.split(" "))
        return Ret_list

# This have shown a simple, easy, and common object-oriented textbook way of treating data with encapsulation and also included 
# a double hierarchy of objects, methods, and the 2D data objects (list of data is the second hierarchical data dimension).  
# Let's start the catalog and show how to make the catalogue object serializable for file storage and analyzable with
# standard programs expecting tabular or column data.
Mail_c = create_catalog()
Mail_c.list_Mailcat_objects()

# The pickle module implements binary protocols for serializing and de-serializing a Python object structures. “Pickling” is the 
# process converting a Python object hierarchy to a byte stream, and “unpickling” is the inverse process of convering a byte stream 
# back into an object hierarchy. Pickling (and unpickling) are also called “serialization” or “flattening”;

# Lets use Python's Pickle module to "serialize" or "flatten" this object structure and to save it to a binary file.
# After creating and saving the file the Mail_c Object will be deleted, and later be recreated from the file.
file = open('Mail_c.obj', 'wb')
pickle.dump(Mail_c, file)
file.close()

del Mail_c

# print(Mail_c) # to show that the object Mail_c no longer exist.

file = open('Mail_c.obj', 'rb')
Mail_c = pickle.load(file)
file.close()
Mail_c.list_Mailcat_objects()
```
